```python
import pandas as pd
import numpy as py
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, HashingVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import PassiveAggressiveClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('news.csv', index_col=None)
```

```python
dataset = dataset.drop('Unnamed: 0', axis=1)
dataset
```

```python
dataset['label'].value_counts
```

```python
indep = dataset['text']
dep = dataset['label']
```

```python
indep
```

```python
dep
```

```python
X_train,X_test,y_train,y_test = train_test_split(indep,dep,test_size=0.30,random_state=53)
```

```python
countvector = CountVectorizer(stop_words="english")
count_train = countvector.fit_transform(X_train)
print(count_train)
count_test = countvector.transform(X_test)
```

```python
#Naive bayes

classifier = MultinomialNB()
classifier.fit(count_train, y_train)

y_predict = classifier.predict(count_test)

accuracyScore = accuracy_score(y_test,y_predict)
confusionMatrix = confusion_matrix(y_test,y_predict,labels=['FAKE','REAL'])
```

```python
accuracyScore
```

```python
confusionMatrix
```

```python
classificationReport = classification_report(y_test,y_predict)
print(classificationReport)
```
